```python
import turtle as t

# 화면 설정
t.setup(width=700, height=600)
t.bgcolor("white")
t.title("파란 돌고래")
t.speed(0)  # 가장 빠른 속도로 그리기

# 1. 돌고래 외곽선 좌표 (점프하는 돌고래 실루엣)
dolphin_points = [
    (170, 20),   # 주둥이 끝
    (148, 25),   # 주둥이-이마 경계
    (135, 38),   # 이마(멜론)
    (115, 52),
    (85, 64),
    (50, 72),
    (20, 76),    # 등지느러미 시작
    (10, 88),
    (0, 106),
    (-10, 122),  # 등지느러미 끝
    (-16, 105),
    (-20, 90),
    (-25, 74),   # 등지느러미 끝단
    (-50, 65),   # 등 라인
    (-80, 52),
    (-115, 36),
    (-145, 18),
    (-170, 0),   # 꼬리 시작점
    (-195, 15),  # 꼬리지느러미 위쪽 날개
    (-220, 26),  # 꼬리 윗끝
    (-212, 14),
    (-198, 3),
    (-186, -6),  # 꼬리 중앙 홈
    (-200, -18),
    (-215, -28),
    (-222, -36), # 꼬리 아랫끝
    (-205, -26),
    (-188, -14),
    (-165, -8),  # 꼬리 아래
    (-130, 2),   # 배 라인
    (-90, 12),
    (-50, 20),
    (-10, 24),
    (15, 22),    # 가슴지느러미 뒤쪽
    (8, 0),
    (0, -25),    # 가슴지느러미 끝
    (12, -15),
    (28, 5),
    (42, 20),    # 가슴지느러미 앞쪽
    (75, 19),    # 턱 라인
    (110, 18),
    (145, 17),   # 아랫입술
    (165, 18),
    (170, 20)    # 주둥이 끝 복귀
]

# 2. 몸통 그리기
t.penup()
t.goto(dolphin_points[0])
t.pendown()

t.color("#1E88E5")  # 선과 채우기 모두 선명한 파란색
t.begin_fill()
for point in dolphin_points:
    t.goto(point)
t.end_fill()

# 3. 눈 그리기
t.penup()
t.goto(120, 32)
t.pendown()
t.color("white")
t.begin_fill()
t.circle(5)
t.end_fill()

# 눈동자
t.penup()
t.goto(122, 34)
t.pendown()
t.color("black")
t.begin_fill()
t.circle(2.5)
t.end_fill()

# 거북이 숨기기 및 완료 유지
t.hideturtle()
t.done()
```
