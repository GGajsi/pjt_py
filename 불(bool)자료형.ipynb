### 불(Bool, Boolean) 자료형

* bool자료형이란 `참과 거짓을 나타내는 자료형`이다.
* bool자료형은 2가지 값인 `True` or `False`값만 가질 수 있다.
* 주의할 점은 True와 False는 파이썬의 예약어로서 첫글자는 반드시 대문자이어야 한다.

```python
a = True
b = False
```

```python
type(a)
type(b)
1 == 1
2 > 1
2 < 1
```

```python
a = [1,2,3,4]
while a :
    print(a.pop())
```

```python
if [1,2,3]:
    print("참")
else:
    print("거짓")
```
