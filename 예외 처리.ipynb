### 예외처리(Exception)

#### 1. 에러와 예외

1. 에러 (Syntax Error) : 문법에러, 파이썬은 상대적으로 문법이 간단하기 때문에 구문자체 에러 즉, 문법에러의 발생비율이 적다.
2. 예외 (Exception) : 예외는 실행중에 발생하는 에러로서 문법에러는 없으나 프로그램이 실행 중에 더 이상 진행할 수 없는 상태를 말한다. 예외가 발생이 되면 프로그램이 바로 중단이 된다. 이런 중단이 되는 상황을 방지하기 위해 예외처리를 해야 한다.

#### 2. 예외처리 방법

> `try:` <br>
> 예외가 발생할 가능성이 있는 실행문자.... (예를 들면 c = b / 0) <br>
> `except Exception:` <br>
> 예외가 발생했을 경우 실행할 문자.... (print('0으로 나눌 수 없습니다.') <br>
> `else:` <br>
> 정상적으로 처리할 경우의 실행 문자 <br>
> `finally:` <br>
> 한번은 반드시 실행해야 할 문자

### try-except 문

```python
try:
    ...
except [발생_오류[as 오류_변수]]:
    ...
```

1, try-except만 쓰는법

```python
try:
    ...
except:
    ...
```

2. 발생 오류만 포함한 expept 문

```python
try:
    ...
except 발생_오류:
    ...
```

3. 발생 오류와 변수까지 포함한 expept 문

```python
try:
    ...
except 발생_오류 as 오류_변수:
    ...
```

```python
try:
    4 / 0
except ZeroDivisionError as e:
    print(e)
```

### try-finally

```python
try:
    f = open('foo.txt', 'w')
    #무언가를 수행

    (...생략...)

finally:
    f.close()   #중간에 오류가 발생 하더라도 무조건 실행
```

### 여러개의 오류 처리하기

```python
try:
    ...
except 발생_오류1:
    ...
except 발생_오류2:
    ...
```

```python
try:
    a = [1,2]
    print(a[3])
    4 / 0
except ZeroDivisionError:
    print("0으로 나눌 수 없습니다.")
except IndexError:
    print("인덱싱할 수 없습니다.")
```

```python
try:
    a = [1,2]
    print(a[3])
    4/0
except ZeroDivisionError as e:
    print(e)
except IndexError as e:
    print(e)
```

```python
try:
    a = [1,2]
    print(a[3])
    4 / 0
except (ZeroDivisionError, IndexError) as e:
    print(e)
```

### try-else 문

```python
try:
    ...
except [발생_오류 [as 오류_변수]]:
    ...
else:           #오류가 없을 경우에만 수행
    ...
```

```python
try:
    age = int(input('나이를 입력하세요: '))
except:
    print('입력이 정확하지 않습니다.')
else:
    if age <= 18:
        print('미셩년자는 출입금지입니다.')
    else:
        print('환영합니다.')
```

### 오류 피하기

```python
try:
    f = open("나없는 파일", 'r')
except FileNotFoundError:       # 파일이 없더라도 오류가 발생하지 않고 통과
    pass
```

### 오류 일부러 발생시키기

```python
class Bird:
    def fly(self):
        raise NotImplementedError
```

```python
class Eagle(Bird):
    pass


eagle = Eagle()
eagle.fly()
```

```python
class Eagle(Bird):
    def fly(self):
        print("very fast")

eagle = Eagle()
eagle.fly()
```

### 예외 만들기

```python
class MyError(Exception):
    def __str__(self):
        return "허용되지 않는 별명입니다."
```

```python
def say_nick(nick):
    if nick == '바보':
        raise MyError()
    print(nick)
```

```python
try:
    say_nick("천사")
    say_nick("바보")
except MyError:
    print("허용되지 않는 병명입니다.")
```

```python
try:
    say_nick("천사")
    say_nick("바보")
except MyError as e:
    print(e)
```
