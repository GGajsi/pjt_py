### 비교 연산자

* `==` : 왼쪽 값과 오른쪽 값이 같음
* `!=` : 왼쪽 값과 오른쪽 값이 다름
* `>` : 왼쪽 값이 오른쪽 값보다 큼 (초과)
* `<` : 왼쪽 값이 오른쪽 값보다 작음 (미만)
* `>=` : 왼쪽 값이 오른쪽 값보다 크거나 같음 (이상)
* `<=` : 왼쪽 값이 오른쪽 값보다 작거나 같음 (이하)

```python
money = 3000
if money >= 3000:
    print("과자 한봉지를 구매")
else:
    print("마이쮸를 구매")
```

### 논리 연산자

* `and` : 두 조건이 모두 참일 때만 참(True)
* `or` : 두 조건 중 하나라도 참이면 참(True)
* `not` : 조건의 결과를 반대로 반전 (참이면 거짓, 거짓이면 참)

```python
money = 2000
card = True     #카드를 가지고 있다고 설정
if money >= 3000 or card:   #만약 3000천원이상 이나 카드를 가지고 있으면.....

    print("버스를 타라")
else:
    print("걸어가라")
```

### 멤버십(포함) 연산자

* `in` : 지정한 값이 시퀀스(리스트, 튜플, 문자열 등) 내에 존재하면 참(True)
* `not in` : 지정한 값이 시퀀스(리스트, 튜플, 문자열 등) 내에 존재하지 않으면 참(True)

```python
1 in [1,2,3]    #1이 있어 True

1 not in [1,2,3]    #1이 있어 False
```

```python
"a" in "apple"  #True

"j" not in "Gojo Satoru"    #Flase

"h" not in "Han JACK"   #True
```

```python
#만약 주너미에 돈이 있으면 택시를 타고 가소, 어뵤으면 걸어가라.
pocket = ['paper', 'cellphone', 'money']
if 'money' in pocket:
    print("택시를 타고가라")
else:
    print("걸어가라")
```

```python
#주머니에 돈이 있으면 가만히 있고, 주머니에 돈이 없으면 카드를 꺼내라.
pocket = ['paper', 'money', 'cellphone']
if 'money' in pocket:
    pass
else:
    print("카드를 꺼내라")
```

```python
#주머니에 돈이 있으면 택시를 타고 가고, 주머니에 돈은 없지만 카드가 있으면 택시를 타고 가고, 돈도 없고 카드도 없으면 걸어가라.
pocket = ['paper', 'cellpheone']
card = True
if 'money' in pocket:
    print("택시를 타고 가라")
else:
    if card:
        print("택시를 타고 가라")
    else:
        print("걸어가라")
```
