```python
class FourCal:
    def setdata(self, first, second):
        self.first = first
        self.second = second

    def add(self):
        result = self.first + self.second  # '*'에서 '+'로 수정
        return result

    def sub(self):
        result = self.first - self.second
        return result

    def div(self):
        result = self.first / self.second
        return result

# 객체 생성 및 데이터 설정
a = FourCal()
b = FourCal()

a.setdata(4, 2)
b.setdata(3, 8)

# 메서드 호출
print(a.add())  # 출력: 6
print(a.sub())  # 출력: 2
print(a.div())  # 출력: 2.0
```

생성자

```python
class FourCal:  # 'lass' -> 'class'로 수정
    def __init__(self, first, second):
        self.first = first
        self.second = second

    def setdata(self, first, second):
        self.first = first
        self.second = second

    def add(self):
        return self.first + self.second

    def mul(self):
        return self.first * self.second

    def sub(self):
        return self.first - self.second

    def div(self):
        return self.first / self.second


# 객체 생성 시 초기값 전달
a = FourCal(4, 2)

print(a.add())  # 6
print(a.div())  # 2.0

a = FourCal(4,2)
a.first

a.add()
```

클래스 상속

```python

class MoreFourCal(FourCal):
    pass  #

a = MoreFourCal(4, 2)

print(a.add())  # 6
print(a.mul())  # 8
print(a.sub())  # 2
print(a.div())  # 2.0
```

```python
class MoreFourCal(FourCal):
    def pow(self):
        result = self.first ** self.second
        return result

a = MoreFourCal(4,2)
a.pow()

a.add()
```

에서드 오버라이딩

```python
class SafeFourCal(FourCal):
    def div(self):
        if self.second == 0:
            return 0
        else:
            return self.first / self.second

a = SafeFourCal(4, 0)

print(a.div())
```
